## Instalando dependências

```python
# --python .venv/Scripts/python.exe faz o uv instalar DENTRO do .venv do projeto
# (sem isso ele instala no Python global, que foi o que apareceu no uv pip list)
!uv pip install --python .venv/Scripts/python.exe pandas
!uv pip list --python .venv/Scripts/python.exe
```

```python
# Congelamos as versoes exatas do ambiente em um arquivo
%cd "C:/Users/USER/Documents/Projetos/Estudos/Ciencia de dados"

# o > grava a saida do comando dentro do arquivo requirements.txt
!uv pip freeze --python .venv/Scripts/python.exe > requirements.txt
# O que ficou registrado? 
!type requirements.txt
```

## Commits das etapas
Cada etapa termina com um commit: primeiro o .gitignore, depois o requirements.txt.

```python
# O .gitignore faz parte do projeto e deve ser versionado
!git add .gitignore
!git commit -m "Adiciona .gitignore para ambiente, dados e segredos"
!git log --oneline
```

```python
# Versionamos a receita (requirements.txt), nao a cozinha (.venv)
!git add requirements.txt
!git commit -m "Adiciona pandas e registra dependencias em requirements.txt"
!git log --oneline
```
